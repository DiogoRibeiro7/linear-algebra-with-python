```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path("../packages/python/src").resolve()))

import numpy as np
from linalg_utils.determinants import det_2x2, det_3x3_sarrus, eh_invertivel

# 2x2 determinants
A = np.array([[1, 2], [3, 4]], dtype=float)
print(f"det_2x2([[1,2],[3,4]]) = {det_2x2(A)}")  # -2

B = np.array([[2, 4], [1, 2]], dtype=float)  # singular
print(f"det_2x2([[2,4],[1,2]]) = {det_2x2(B)}")  # 0

# 3x3 determinants via Sarrus
C = np.array([[1, 2, 3], [0, 1, 4], [5, 6, 0]], dtype=float)
print(f"\ndet_3x3_sarrus = {det_3x3_sarrus(C)}")  # 1

# Invertibility check
print(f"\nA invertible: {eh_invertivel(A)}")  # True
print(f"B invertible: {eh_invertivel(B)}")    # False
```

```python
from linalg_utils.determinants import det_gauss

# det_gauss with elimination steps
A4 = np.array([
    [2, 1, -1, 3],
    [4, 5, -3, 7],
    [-2, 3, 1, -1],
    [6, -2, 4, 5],
], dtype=float)

det_val, steps = det_gauss(A4)
print(f"det(A4) via Gauss = {det_val:.6f}")
print(f"det(A4) via NumPy  = {np.linalg.det(A4):.6f}")
print(f"Number of elimination steps: {len(steps)}")

for i, step in enumerate(steps):
    print(f"\nStep {i+1}:\n{step}")
```

```python
from linalg_utils.determinants import det_cofactor

# Verify determinant properties
A = np.array([[2, 1], [3, 4]], dtype=float)
B = np.array([[1, 5], [2, 3]], dtype=float)

# Property 1: det(AB) = det(A) * det(B)
print(f"det(A)*det(B) = {det_cofactor(A) * det_cofactor(B):.6f}")
print(f"det(AB)       = {det_cofactor(A @ B):.6f}")
assert np.isclose(det_cofactor(A @ B), det_cofactor(A) * det_cofactor(B))

# Property 2: det(A^T) = det(A)
print(f"\ndet(A)   = {det_cofactor(A):.6f}")
print(f"det(A^T) = {det_cofactor(A.T):.6f}")
assert np.isclose(det_cofactor(A), det_cofactor(A.T))

# Property 3: det(kA) = k^n * det(A) for nxn matrix
k = 3.0
n = A.shape[0]
print(f"\ndet(3A)      = {det_cofactor(k * A):.6f}")
print(f"3^2 * det(A) = {k**n * det_cofactor(A):.6f}")
assert np.isclose(det_cofactor(k * A), k**n * det_cofactor(A))

print("\nAll determinant properties verified.")
```

```python
# eh_invertivel with condition numbers
matrices = {
    "Well-conditioned": np.array([[2, 1], [1, 3]], dtype=float),
    "Ill-conditioned": np.array([[1, 1], [1, 1 + 1e-10]], dtype=float),
    "Singular": np.array([[1, 2], [2, 4]], dtype=float),
}

for name, M in matrices.items():
    inv = eh_invertivel(M)
    cond = np.linalg.cond(M)
    print(f"{name}: invertible={inv}, cond={cond:.2e}")
```

```python
from linalg_utils.determinants import inversa_gauss_jordan

A = np.array([[2, 1, 0], [3, 4, 1], [1, 0, 2]], dtype=float)
A_inv, steps = inversa_gauss_jordan(A)

print(f"A:\n{A}")
print(f"\nA^(-1):\n{A_inv}")
print(f"\nA @ A^(-1):\n{A @ A_inv}")

assert np.allclose(A @ A_inv, np.eye(3), atol=1e-9)
print("\nVerified: A @ A^(-1) = I")
```

```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from linalg_utils.geometry2d import unit_square, apply_linear_map

A = np.array([[2, 1], [0.5, 1.5]], dtype=float)
pts = unit_square(100)
transformed = apply_linear_map(A, pts)
det_A = np.linalg.det(A)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(pts[:, 0], pts[:, 1], "b-")
axes[0].set_title("Unit square")
axes[0].set_aspect("equal")
axes[0].grid(True)

axes[1].plot(transformed[:, 0], transformed[:, 1], "r-")
axes[1].set_title(f"A * unit square, |det(A)| = {abs(det_A):.2f}")
axes[1].set_aspect("equal")
axes[1].grid(True)

plt.tight_layout()
plt.savefig("determinante_area.png", dpi=120, bbox_inches="tight")
plt.close()
print("Saved determinante_area.png")
```
