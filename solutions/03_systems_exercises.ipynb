# Systems exercises — solutions
Answers and commentary for each task.


```python
import numpy as np
A = np.array([[3.0, 1.0], [1.0, 2.0]])
b = np.array([9.0, 8.0])
x = np.linalg.solve(A, b)
print('Solution:', x)
```

```python
residual = A @ x - b
print('Residual norm:', np.linalg.norm(residual))
```

```python
print('Condition number:', np.linalg.cond(A))
```

```python
delta = 1e-3
b_perturbed = b + np.array([delta, -delta])
x_pert = np.linalg.solve(A, b_perturbed)
print('Perturbed change:', np.linalg.norm(x_pert - x))
```

```python
A_bad = np.array([[1.0, 1.0], [1.0, 1.001]])
b_bad = np.array([2.0, 2.001])
x_bad, *_ = np.linalg.lstsq(A_bad, b_bad, rcond=None)
print('Ill-conditioned solution:', x_bad)
```

```python
deltas = np.logspace(-6, -1, 5)
errors = []
for eps in deltas:
    b_eps = b + np.array([eps, -eps])
    x_eps = np.linalg.solve(A, b_eps)
    errors.append(np.linalg.norm(x_eps - x))
print('Errors:', errors)
```

```python
A_norm = np.linalg.norm(A, 2)
A_inv_norm = np.linalg.norm(np.linalg.inv(A), 2)
print('Matrix norm:', A_norm)
print('Inverse norm:', A_inv_norm)
```

## Check your work

- Perturbation sweeps show solution sensitivity increases with ill-conditioning.
