```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path("../packages/python/src").resolve()))

import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from linalg_utils.geometry import superficie_cilindro, superficie_cone, superficie_esfera

# Standard surfaces
fig = plt.figure(figsize=(15, 5))

for i, (func, title) in enumerate([
    (superficie_cilindro, "Cylinder (r=1, h=2)"),
    (superficie_cone, "Cone (r=1, h=2)"),
    (superficie_esfera, "Sphere (r=1)"),
]):
    X, Y, Z = func()
    ax = fig.add_subplot(1, 3, i+1, projection="3d")
    ax.plot_surface(X, Y, Z, alpha=0.6, cmap="viridis")
    ax.set_title(title)
    ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")

plt.tight_layout()
plt.savefig("superficies_padrao.png", dpi=120, bbox_inches="tight")
plt.close()
print("Saved superficies_padrao.png")
```

```python
# Sphere cross-sections for radius 2
r = 2
theta = np.linspace(0, 2 * np.pi, 100)

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

# Full sphere
X, Y, Z = superficie_esfera(r=r, n=30)
ax.plot_surface(X, Y, Z, alpha=0.15, color="cyan")

# Cross-sections at z = -1, 0, 1
for z_val in [-1, 0, 1]:
    r_cross = np.sqrt(r**2 - z_val**2)
    ax.plot(r_cross * np.cos(theta), r_cross * np.sin(theta),
            z_val * np.ones_like(theta), linewidth=2,
            label=f"z={z_val}, r={r_cross:.2f}")

ax.set_title(f"Sphere (r={r}) with cross-sections")
ax.legend()
plt.tight_layout()
plt.savefig("esfera_cortes.png", dpi=120, bbox_inches="tight")
plt.close()
print("Saved esfera_cortes.png")
```

```python
from linalg_utils.geometry import superficie_revolucao

# Wavy vase: r(t) = 1 + 0.3*sin(4t), z(t) = t
def wavy_vase(t):
    r = 1 + 0.3 * np.sin(4 * t)
    z = t
    return r, z

# Paraboloid: r(t) = sqrt(t), z(t) = t
def paraboloid(t):
    r = np.sqrt(t)
    z = t
    return r, z

fig = plt.figure(figsize=(12, 5))

for i, (func, title, t_range) in enumerate([
    (wavy_vase, "Wavy Vase", (0, 3)),
    (paraboloid, "Paraboloid", (0.01, 3)),
]):
    X, Y, Z = superficie_revolucao(func, eixo="z", t_range=t_range, n=60)
    ax = fig.add_subplot(1, 2, i+1, projection="3d")
    ax.plot_surface(X, Y, Z, alpha=0.6, cmap="plasma")
    ax.set_title(title)
    ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")

plt.tight_layout()
plt.savefig("superficies_revolucao.png", dpi=120, bbox_inches="tight")
plt.close()
print("Saved superficies_revolucao.png")
```

```python
# Gallery: cylinder, cone, sphere, wavy vase, paraboloid, hyperboloid
def hyperboloid(t):
    r = np.sqrt(1 + t**2)
    z = t
    return r, z

surfaces = [
    ("Cylinder", superficie_cilindro()),
    ("Cone", superficie_cone()),
    ("Sphere", superficie_esfera()),
    ("Wavy Vase", superficie_revolucao(wavy_vase, eixo="z", t_range=(0, 3), n=50)),
    ("Paraboloid", superficie_revolucao(paraboloid, eixo="z", t_range=(0.01, 3), n=50)),
    ("Hyperboloid", superficie_revolucao(hyperboloid, eixo="z", t_range=(-2, 2), n=50)),
]

fig = plt.figure(figsize=(18, 10))
for i, (title, (X, Y, Z)) in enumerate(surfaces):
    ax = fig.add_subplot(2, 3, i+1, projection="3d")
    ax.plot_surface(X, Y, Z, alpha=0.6, cmap="viridis")
    ax.set_title(title)
    ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")

plt.suptitle("Gallery of Surfaces of Revolution", fontsize=14)
plt.tight_layout()
plt.savefig("galeria_superficies.png", dpi=120, bbox_inches="tight")
plt.close()
print("Saved galeria_superficies.png")
```
