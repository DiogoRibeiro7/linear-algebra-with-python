```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path("../packages/python/src").resolve()))

import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from linalg_utils.geometry import posicao_relativa_retas_2d, reta_2d

# posicao_relativa_retas_2d — 3 cases
cases_2d = [
    ("Intersecting", [0,0], [1,1], [0,1], [1,-1]),
    ("Parallel", [0,0], [1,0], [0,1], [1,0]),
    ("Coincident", [0,0], [1,1], [1,1], [2,2]),
]

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, (label, p1, d1, p2, d2) in zip(axes, cases_2d):
    result = posicao_relativa_retas_2d(p1, d1, p2, d2)
    x1, y1 = reta_2d(p1, d1, t_range=(-2, 2))
    x2, y2 = reta_2d(p2, d2, t_range=(-2, 2))
    ax.plot(x1, y1, "b-", label="Line 1")
    ax.plot(x2, y2, "r--", label="Line 2")
    ax.set_title(f"{label}: {result}")
    ax.legend()
    ax.grid(True)
    ax.set_aspect("equal")

plt.tight_layout()
plt.savefig("retas_2d.png", dpi=120, bbox_inches="tight")
plt.close()
print("Saved retas_2d.png")
```

```python
# Dot product tests for parallelism and perpendicularity
a = np.array([1, 2, 3], dtype=float)
b = np.array([2, 4, 6], dtype=float)  # parallel (b = 2a)
c = np.array([-2, 1, 0], dtype=float) # perpendicular to a

print(f"a . b = {np.dot(a, b)} (parallel: cross product = {np.cross(a, b)})")
print(f"a . c = {np.dot(a, c)} (perpendicular: dot product is 0)")

# General test
def check_relation(u, v):
    cross = np.cross(u, v)
    dot = np.dot(u, v)
    if np.allclose(cross, 0):
        return "parallel"
    elif np.isclose(dot, 0):
        return "perpendicular"
    return "general"

print(f"\na,b: {check_relation(a, b)}")
print(f"a,c: {check_relation(a, c)}")
print(f"a,[1,1,1]: {check_relation(a, [1,1,1])}")
```
