```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path("../packages/python/src").resolve()))

import numpy as np
from linalg_utils.matrices import dimensao

# Define matrices
A = np.array([[1, 2], [3, 4], [5, 6]], dtype=float)   # 3x2
B = np.array([[7, 8], [9, 10], [11, 12]], dtype=float) # 3x2

# Addition and subtraction (same dimensions)
print("A + B:\n", A + B)
print("\nA - B:\n", A - B)
print("\n3A:\n", 3 * A)
print(f"\ndim(A)={dimensao(A)}, dim(B)={dimensao(B)}, dim(A+B)={dimensao(A + B)}")
```

```python
# Multiplication with dimension analysis
A = np.array([[1, 2, 3], [4, 5, 6]], dtype=float)   # 2x3
C = np.array([[1, 0], [0, 1], [1, 1]], dtype=float)  # 3x2

AC = A @ C  # (2x3) @ (3x2) = 2x2
print(f"A(2x3) @ C(3x2) = AC(2x2):\n{AC}")

CA = C @ A  # (3x2) @ (2x3) = 3x3
print(f"\nC(3x2) @ A(2x3) = CA(3x3):\n{CA}")

print(f"\nAC != CA: multiplication is not commutative")
print(f"dim(AC)={dimensao(AC)}, dim(CA)={dimensao(CA)}")
```

```python
# Commutator [P, Q] = PQ - QP and identity behavior
P = np.array([[1, 2], [3, 4]], dtype=float)
Q = np.array([[5, 6], [7, 8]], dtype=float)
I = np.eye(2)

commutator = P @ Q - Q @ P
print(f"[P, Q] = PQ - QP:\n{commutator}")
print(f"\nCommutator is zero: {np.allclose(commutator, 0)}")

# Identity: [P, I] should be zero
comm_identity = P @ I - I @ P
print(f"\n[P, I]:\n{comm_identity}")
print(f"[P, I] is zero: {np.allclose(comm_identity, 0)} (identity commutes with everything)")
```

```python
# Verify transpose properties
A = np.array([[1, 2, 3], [4, 5, 6]], dtype=float)
B = np.array([[7, 8, 9], [10, 11, 12]], dtype=float)
k = 3.0

# (A^T)^T = A
assert np.allclose(A.T.T, A), "(A^T)^T should equal A"

# (A + B)^T = A^T + B^T
assert np.allclose((A + B).T, A.T + B.T), "(A+B)^T should equal A^T + B^T"

# (kA)^T = k(A^T)
assert np.allclose((k * A).T, k * A.T), "(kA)^T should equal k*A^T"

# (AB)^T = B^T A^T
C = np.array([[1, 0], [0, 1], [1, 1]], dtype=float)
assert np.allclose((A @ C).T, C.T @ A.T), "(AC)^T should equal C^T @ A^T"

print("All transpose properties verified.")
```

```python
# Ax as linear combination of A's columns
A = np.array([[1, 2], [3, 4], [5, 6]], dtype=float)
x = np.array([2, 3], dtype=float)

# Direct multiplication
Ax = A @ x
print(f"A @ x = {Ax}")

# As linear combination: x[0]*col_0 + x[1]*col_1
col_0 = A[:, 0]
col_1 = A[:, 1]
lin_comb = x[0] * col_0 + x[1] * col_1
print(f"x[0]*A[:,0] + x[1]*A[:,1] = {lin_comb}")

assert np.allclose(Ax, lin_comb)
print("\nAx equals the linear combination of A's columns weighted by x.")
```
