Filled answers for each task with explanations and expected behavior.


```python
import numpy as np
from linalg_utils import geometry2d

transform = (
    np.array([[np.cos(np.pi / 6), -np.sin(np.pi / 6)], [np.sin(np.pi / 6), np.cos(np.pi / 6)]])
    @ np.diag([1.2, 1.0])
)
v = np.array([1.0, 0.5])
mapped = transform @ v
print('Original:', v)
print('Mapped:', mapped)
```

```python
basis = np.eye(2)
mapped_basis = geometry2d.apply_linear_map(transform, basis)
print('Dot product between mapped basis:', np.dot(mapped_basis[0], mapped_basis[1]))
```

```python
square = geometry2d.unit_square(120)
square_mapped = geometry2d.apply_linear_map(transform, square)
print('First corner map:', square[0], '->', square_mapped[0])
```

```python
circle = geometry2d.unit_circle()
circle_mapped = geometry2d.apply_linear_map(transform, circle)
print('Max radius after map:', np.max(np.linalg.norm(circle_mapped, axis=1)))
```

```python
shear = np.array([[1.0, 0.5], [0.0, 1.0]])
composed = shear @ transform
inv_composed = np.linalg.inv(composed)
recovered = geometry2d.apply_linear_map(inv_composed, square_mapped)
print('Recovered first vertex close to original:', np.allclose(recovered[0], square[0]))
```

```python
candidate = np.array([1.0, 1.0])
mapped_candidate = geometry2d.apply_linear_map(transform, candidate[np.newaxis, :])[0]
print('Stretch factor:', np.linalg.norm(mapped_candidate) / np.linalg.norm(candidate))
```

```python
square = geometry2d.unit_square(120)
mapped_square = geometry2d.apply_linear_map(transform, square)
manual = (transform @ square.T).T
print('Manual difference:', np.max(np.abs(manual - mapped_square)))
```

```python
reflection = np.diag([1.0, -1.0])
circle = geometry2d.unit_circle()
reflected = geometry2d.apply_linear_map(reflection, circle)
print('Y min before vs after:', circle[:, 1].min(), reflected[:, 1].min())
```

## Check your work
- Mapped basis dot product should stay near zero to confirm linear independence.
- The square is closed and recovers under inversion, meaning composed maps behave geometrically.
- Stretch factors show which directions dominate the map.
