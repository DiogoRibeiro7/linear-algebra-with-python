## Track A — QR least squares on noisy line data
We generate noisy points around `y = 2x + 1` and use `least_squares_qr`. The residual projection is visualized alongside the fitted line.


```python
import numpy as np
import matplotlib.pyplot as plt
from linalg_with_python.least_squares import least_squares_qr
from linalg_with_python import checks

rng = np.random.default_rng(0)
x = np.linspace(0, 5, 30)
y_true = 2 * x + 1
y = y_true + rng.normal(scale=0.5, size=x.shape)
A = np.vstack([x, np.ones_like(x)]).T
b = y
result = least_squares_qr(A, b)
fit = A @ result.x
residual = b - fit
residual_proj = A @ (np.zeros_like(result.x))
print('Residual norm:', result.residual_norm)
print('Check orthonormal basis (QR guarantees):', checks.is_orthonormal(np.linalg.qr(A, mode='reduced')[0]))
fig, ax = plt.subplots()
ax.scatter(x, b, label='Noisy points')
ax.plot(x, fit, label='QR fit', color='k')
ax.plot(x, result.x[0] * x + result.x[1], '--', label='Line eqn')
ax.set_title('Track A fit and residual')
ax.legend()
fig.savefig('assets/figures/capstone_trackA.png')
```

## Track B — PCA / SVD on correlated data
SVD extracts the principal directions; projecting onto the first component recovers most variance.


```python
X = rng.multivariate_normal([0, 0], [[4, 3.5], [3.5, 4]], size=200)
X_centered = X - X.mean(axis=0)
U, S, Vt = np.linalg.svd(X_centered, full_matrices=False)
first_pc = Vt[0]
projected = X_centered @ first_pc[:, None] * first_pc
explained_variance = S**2 / (len(X) - 1)
print('First eigenvector:', first_pc)
print('Explained variance ratio (first PC):', explained_variance[0] / explained_variance.sum())
plt.figure()
plt.scatter(X_centered[:, 0], X_centered[:, 1], alpha=0.5, label='Data')
plt.quiver(0, 0, first_pc[0], first_pc[1], color='r', scale=3, label='1st PC')
plt.title('Track B: PCA direction')
plt.legend()
plt.savefig('assets/figures/capstone_trackB.png')
```
