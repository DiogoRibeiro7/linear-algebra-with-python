# Assignment 2 — Matrix Operations and Property Verification
This notebook explores basic matrix operations and verifies key algebraic properties computationally.

## 1. Basic Operations
We compute addition, subtraction, and scalar multiplication.

```
import numpy as np

# Define matrices
A = np.array([[1, 2, 3], [4, 5, 6]])       # 2×3
B = np.array([[7, 8, 9], [10, 11, 12]])     # 2×3

print("A ="); print(A)
print("\nB ="); print(B)
print(f"\nDimensions of A: {A.shape[0]}×{A.shape[1]}")
print(f"Dimensions of B: {B.shape[0]}×{B.shape[1]}")

# Sum and difference
print("\n--- Sum A + B ---"); print(A + B)
print("\n--- Difference A - B ---"); print(A - B)
print("\n--- Scalar product 3A ---"); print(3 * A)

# Verify: (A + B) has same shape as A
assert (A + B).shape == A.shape, "Incompatible dimensions!"
print("\n✓ The sum preserves the dimensions.")
```

## 2. Matrix Product
Matrix multiplication requires cols(A) = rows(B). The result has shape rows(A) × cols(B).

```
C = np.array([[1, 2], [3, 4], [5, 6]])  # 3×2

AC = A @ C   # (2×3)(3×2) = 2×2
print("A (2×3) @ C (3×2) ="); print(AC)
print(f"Dimensions of AC: {AC.shape}")

# Can we compute CA?
CA = C @ A   # (3×2)(2×3) = 3×3
print("\nC (3×2) @ A (2×3) ="); print(CA)
print(f"Dimensions of CA: {CA.shape}")

print("\n⚠ AC has dimension 2×2 but CA has dimension 3×3 — they are different matrices!")
print("Even when both products exist, generally AC ≠ CA.")
```

## 3. Non-commutativity of the Product
For square matrices, PQ and QP both exist but are generally different. The commutator measures this difference.

```
P = np.array([[1, 2], [3, 4]])
Q = np.array([[5, 6], [7, 8]])

PQ = P @ Q
QP = Q @ P
commutator = PQ - QP

print("P ="); print(P)
print("\nQ ="); print(Q)
print("\nPQ ="); print(PQ)
print("\nQP ="); print(QP)
print("\nCommutator [P, Q] = PQ - QP ="); print(commutator)
print(f"\nPQ == QP? {np.allclose(PQ, QP)}")
print("Matrix multiplication is NOT commutative in general.")

# Show a case where it IS commutative: A @ I = I @ A
I = np.eye(2)
print(f"\nBut P @ I == I @ P? {np.allclose(P @ I, I @ P)} (the identity commutes with all matrices)")
```

## 4. Transpose Properties
We verify: (Aᵀ)ᵀ = A, (A+B)ᵀ = Aᵀ+Bᵀ, and (AC)ᵀ = CᵀAᵀ.

```
print("=== Property 1: (Aᵀ)ᵀ = A ===")
print(f"(Aᵀ)ᵀ == A? {np.allclose(A.T.T, A)}")

print("\n=== Property 2: (A + B)ᵀ = Aᵀ + Bᵀ ===")
lhs = (A + B).T
rhs = A.T + B.T
print(f"(A + B)ᵀ == Aᵀ + Bᵀ? {np.allclose(lhs, rhs)}")

print("\n=== Property 3: (AC)ᵀ = Cᵀ Aᵀ ===")
lhs = (A @ C).T
rhs = C.T @ A.T
print(f"(AC)ᵀ == CᵀAᵀ? {np.allclose(lhs, rhs)}")
print("Note: the order reverses in the transpose of a product!")
```

## 5. Symmetric Matrices
A matrix S is symmetric if S = Sᵀ. Any matrix M generates a symmetric matrix via S = M + Mᵀ.

```
np.random.seed(42)
M = np.random.randint(1, 10, (3, 3))
S = M + M.T

print("M (random) ="); print(M)
print("\nS = M + Mᵀ ="); print(S)
print(f"\nIs S symmetric? S == Sᵀ? {np.allclose(S, S.T)}")
print("\nElement-by-element check:")
print("S ="); print(S)
print("Sᵀ ="); print(S.T)
```

## 6. Matrix-Vector Product as a Linear Combination
Ax equals a linear combination of A's columns with coefficients from x.

```
A_sys = np.array([[2, 1, 3], [4, -1, 2], [1, 5, -1]])
x = np.array([1, 2, 3])
b = A_sys @ x

print("A ="); print(A_sys)
print(f"\nx = {x}")
print(f"\nAx = {b}")

# Show as linear combination of columns
col1, col2, col3 = A_sys[:, 0], A_sys[:, 1], A_sys[:, 2]
lincomb = x[0] * col1 + x[1] * col2 + x[2] * col3
print(f"\nLinear combination: {x[0]}·a₁ + {x[1]}·a₂ + {x[2]}·a₃ = {lincomb}")
print(f"\nAx == linear combination? {np.allclose(b, lincomb)}")
print("\n✓ Ax is a linear combination of A's columns with weights given by x.")
```

## Conclusion
Matrix operations follow strict dimension rules, multiplication is generally non-commutative, transpose reverses the order in products, and matrix-vector products are linear combinations of columns.
