# Assignment 7 — Surfaces of Revolution
This notebook builds and visualizes surfaces of revolution in 3D.

## 1. Cylinder

```
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401
from linalg_utils.geometry import superficie_cilindro, superficie_cone, superficie_esfera, superficie_revolucao

X, Y, Z = superficie_cilindro(r=1.5, h=3, n=50)
fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, alpha=0.6)
ax.set_title('Cylinder')
plt.savefig('assets/figures/cilindro.png', dpi=150)
plt.close()
```

## 2. Cone

```
X, Y, Z = superficie_cone(r=2, h=3, n=50)
fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, alpha=0.6)
ax.set_title('Cone')
plt.savefig('assets/figures/cone.png', dpi=150)
plt.close()
```

## 3. Sphere

```
X, Y, Z = superficie_esfera(r=2, n=50)
fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, alpha=0.6)
ax.set_title('Sphere')
plt.savefig('assets/figures/esfera.png', dpi=150)
plt.close()
```

## 4. Generating Curves and Axes

```
summary = [
    ('Cylinder', 'Line parallel to z-axis', 'z', 'x^2 + y^2 = r^2'),
    ('Cone', 'Line through the origin', 'z', 'x^2 + y^2 = (rz/h)^2'),
    ('Sphere', 'Semicircle', 'z', 'x^2 + y^2 + z^2 = r^2'),
]
for row in summary:
    print(row)
```

## 5. Cross-Sections

```
# Sphere cross-sections
r = 2
z_cuts = [0, r/2, r]
theta = np.linspace(0, 2*np.pi, 200)
fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(111, projection='3d')
X, Y, Z = superficie_esfera(r=r, n=50)
ax.plot_surface(X, Y, Z, alpha=0.2)
for zc in z_cuts:
    rc = np.sqrt(max(r**2 - zc**2, 0))
    ax.plot(rc*np.cos(theta), rc*np.sin(theta), zc*np.ones_like(theta))
ax.set_title('Sphere Cross-Sections')
plt.savefig('assets/figures/seccoes_comparacao.png', dpi=150)
plt.close()
```

## 6. Custom Surface of Revolution

```
def curve(t):
    return 1 + 0.3*np.sin(3*t), t

X, Y, Z = superficie_revolucao(curve, eixo='z', t_range=(0, 3), n=100)
fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, alpha=0.6)
ax.set_title('Wavy Vase')
plt.savefig('assets/figures/superficie_personalizada.png', dpi=150)
plt.close()

# Paraboloid of revolution: y = sqrt(x) in x in [0,4]
def parabola(t):
    return np.sqrt(t), t

X, Y, Z = superficie_revolucao(parabola, eixo='z', t_range=(0, 4), n=100)
fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, alpha=0.6)
ax.set_title('Paraboloid of Revolution')
plt.savefig('assets/figures/paraboloide.png', dpi=150)
plt.close()
```

## 7. Final Gallery

```
fig = plt.figure(figsize=(10, 6))
titles = ['Cylinder', 'Cone', 'Sphere', 'Wavy Vase', 'Paraboloid', 'Hyperboloid']
surfaces = []
surfaces.append(superficie_cilindro(r=1.5, h=3, n=30))
surfaces.append(superficie_cone(r=2, h=3, n=30))
surfaces.append(superficie_esfera(r=2, n=30))
surfaces.append(superficie_revolucao(lambda t: (1 + 0.3*np.sin(3*t), t), eixo='z', t_range=(0, 3), n=30))
surfaces.append(superficie_revolucao(lambda t: (np.sqrt(t), t), eixo='z', t_range=(0, 4), n=30))
surfaces.append(superficie_revolucao(lambda t: (np.sqrt(1 + t**2), t), eixo='z', t_range=(0, 2), n=30))

for i, (X, Y, Z) in enumerate(surfaces, start=1):
    ax = fig.add_subplot(2, 3, i, projection='3d')
    ax.plot_surface(X, Y, Z, alpha=0.6)
    ax.set_title(titles[i-1])
    ax.set_axis_off()

plt.tight_layout()
plt.savefig('assets/figures/galeria_superficies.png', dpi=150)
plt.close()
```
