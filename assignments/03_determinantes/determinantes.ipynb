# Assignment 3 — Determinants and Invertibility
This notebook implements and verifies determinant methods and matrix inverses.

## 1. 2×2 Determinant

```
import numpy as np
from linalg_utils.determinants import det_2x2, det_3x3_sarrus, det_gauss, det_cofactor, eh_invertivel, inversa_2x2, inversa_gauss_jordan

A1 = np.array([[2, 3], [1, 4]], dtype=float)
A2 = np.array([[1, 2], [2, 4]], dtype=float)
A3 = np.array([[-1, 2], [3, -4]], dtype=float)

for idx, A in enumerate([A1, A2, A3], start=1):
    det_manual = det_2x2(A)
    det_np = np.linalg.det(A)
    print(f"A{idx} =
{A}")
    print(f"det(A{idx}) manual = {det_manual:.4f}, numpy = {det_np:.4f}")
    print("-" * 40)
```

## 2. 3×3 Determinant — Sarrus' Rule

```
B1 = np.array([[1, 2, 3], [0, 1, 4], [5, 6, 0]], dtype=float)
B2 = np.array([[2, 0, 1], [3, 1, 2], [4, 0, 1]], dtype=float)

for idx, B in enumerate([B1, B2], start=1):
    det_sarrus = det_3x3_sarrus(B)
    det_np = np.linalg.det(B)
    print(f"B{idx} =
{B}")
    print(f"det(B{idx}) Sarrus = {det_sarrus:.4f}, numpy = {det_np:.4f}")
    print("-" * 40)
```

## 3. Determinant by Gaussian Elimination

```
C = np.array([[2, 1, 0, 3], [1, 2, 1, 0], [3, 0, 1, 2], [0, 1, 4, 1]], dtype=float)
det_val, steps = det_gauss(C)
print("Initial matrix:
", C)
for i, step in enumerate(steps, start=1):
    print(f"Step {i}:
{step}")
print(f"det(C) via Gauss = {det_val:.6f}")
print(f"det(C) via NumPy = {np.linalg.det(C):.6f}")
```

## 4. Determinant Properties

```
rng = np.random.default_rng(0)
A = rng.normal(size=(3, 3))
B = rng.normal(size=(3, 3))
AB = A @ B
print(f"det(AB) = {np.linalg.det(AB):.6f}")
print(f"det(A)*det(B) = {(np.linalg.det(A)*np.linalg.det(B)):.6f}")
print(f"Equal? {np.allclose(np.linalg.det(AB), np.linalg.det(A)*np.linalg.det(B))}")

print(f"det(Aᵀ) = {np.linalg.det(A.T):.6f}, det(A) = {np.linalg.det(A):.6f}")
print(f"Equal? {np.allclose(np.linalg.det(A.T), np.linalg.det(A))}")

k = 3
print(f"det(kA) = {np.linalg.det(k*A):.6f}, k^n det(A) = {(k**3 * np.linalg.det(A)):.6f}")

D = np.array([[1, 2, 3], [2, 4, 6], [0, 1, 2]], dtype=float)
print(f"det(D) = {np.linalg.det(D):.6f} (singular if 0)")

E = A.copy()
E[[0, 1]] = E[[1, 0]]
print(f"Row swap changes sign? {np.allclose(np.linalg.det(E), -np.linalg.det(A))}")

F = A.copy()
F[1] = F[1] + 2 * F[0]
print(f"Row addition preserves det? {np.allclose(np.linalg.det(F), np.linalg.det(A))}")
```

## 5. Invertibility

```
matrices = {
    "invertible": np.array([[2, 1], [1, 3]], dtype=float),
    "singular": np.array([[1, 2], [2, 4]], dtype=float),
    "ill_conditioned": np.array([[1, 1], [1, 1.000001]], dtype=float),
}

for name, M in matrices.items():
    inv = eh_invertivel(M)
    cond = np.linalg.cond(M) if inv else np.inf
    print(f"{name}: invertible={inv}, cond={cond:.2e}")
```

## 6. 2×2 Inverse

```
A = np.array([[2, 1], [5, 3]], dtype=float)
invA = inversa_2x2(A)
print("A =
", A)
print("A^{-1} =
", invA)
print("A @ A^{-1} =
", A @ invA)
```

## 7. Gauss-Jordan Inverse (3×3 and 4×4)

```
G3 = np.array([[1, 2, 3], [0, 1, 4], [5, 6, 0]], dtype=float)
inv3, steps3 = inversa_gauss_jordan(G3)
print("Initial augmented [A|I] for 3x3:
", steps3[0])
print("Final inverse 3x3:
", inv3)
print("Check A @ A^{-1} = I:
", G3 @ inv3)

G4 = np.array([[2, 1, 0, 3], [1, 2, 1, 0], [3, 0, 1, 2], [0, 1, 4, 1]], dtype=float)
inv4, steps4 = inversa_gauss_jordan(G4)
print("Initial augmented [A|I] for 4x4:
", steps4[0])
print("Final inverse 4x4 (showing check):
", G4 @ inv4)
```

## 8. Visualization

```
import matplotlib.pyplot as plt

A = np.array([[1.5, 0.5], [0.2, 1.2]])
square = np.array([[0, 1, 1, 0, 0], [0, 0, 1, 1, 0]], dtype=float)
transformed = A @ square

plt.figure(figsize=(5, 5))
plt.plot(square[0], square[1], label='Unit square')
plt.plot(transformed[0], transformed[1], label='Transformed')
plt.axis('equal')
plt.title(f'Determinant as area scale: det(A) = {np.linalg.det(A):.2f}')
plt.legend()
plt.savefig('assets/figures/determinante_area.png', dpi=150)
plt.close()
```
