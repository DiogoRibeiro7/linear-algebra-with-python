## Introduction — The scenario

```
import numpy as np
from linalg_utils.determinants import det_3x3_sarrus, inversa_gauss_jordan
from linalg_utils.systems import resolver_gauss
from linalg_utils.geometry import plano_parametrico

A = np.array([[2, 3, 1], [1, 2, 3], [3, 1, 2]], dtype=float)
b = np.array([20, 22, 18], dtype=float)
print('A =
', A)
print('b =', b)
print('det(A) =', np.linalg.det(A))
```

## 1. Problem Formulation

```
print('System:')
print('2x1 + 3x2 + x3 = 20')
print('x1 + 2x2 + 3x3 = 22')
print('3x1 + x2 + 2x3 = 18')
```

## 2. Solution by Gaussian Elimination

```
x_gauss, classification, steps = resolver_gauss(A, b)
print('Classification:', classification)
print('Solution (Gauss):', x_gauss)
print('Check A @ x:', A @ x_gauss)
```

## 3. Solution by Cramer's Rule

```
detA = det_3x3_sarrus(A)
A1 = A.copy(); A1[:, 0] = b
A2 = A.copy(); A2[:, 1] = b
A3 = A.copy(); A3[:, 2] = b
x1 = det_3x3_sarrus(A1) / detA
x2 = det_3x3_sarrus(A2) / detA
x3 = det_3x3_sarrus(A3) / detA
x_cramer = np.array([x1, x2, x3])
print('Solution (Cramer):', x_cramer)
print('Check A @ x:', A @ x_cramer)
```

## 4. Solution by Matrix Inverse

```
A_inv, steps_inv = inversa_gauss_jordan(A)
x_inv = A_inv @ b
print('A^{-1} =
', A_inv)
print('Solution (Inverse):', x_inv)
print('Check A @ A^{-1} =
', A @ A_inv)
```

## 5. Comparison of the Three Methods

```
import timeit

def time_fn(fn):
    return min(timeit.repeat(fn, repeat=3, number=1000)) / 1000

t_gauss = time_fn(lambda: resolver_gauss(A, b)[0])
t_cramer = time_fn(lambda: x_cramer)
t_inv = time_fn(lambda: A_inv @ b)

print('Gauss:', x_gauss, 'time', t_gauss)
print('Cramer:', x_cramer, 'time', t_cramer)
print('Inverse:', x_inv, 'time', t_inv)
```

## 6. Sensitivity Analysis

```
rng = np.random.default_rng(0)
condA = np.linalg.cond(A)
print('cond(A) =', condA)

deltas = []
responses = []
for _ in range(10):
    delta = rng.normal(size=3)
    delta = delta / np.linalg.norm(delta)
    b_pert = b + delta
    x_pert, _, _ = resolver_gauss(A, b_pert)
    deltas.append(np.linalg.norm(delta))
    responses.append(np.linalg.norm(x_pert - x_gauss))

import matplotlib.pyplot as plt
plt.figure(figsize=(5, 4))
plt.scatter(deltas, responses)
plt.xlabel('||δ||')
plt.ylabel('||Δx||')
plt.title('Sensitivity')
plt.savefig('assets/figures/sensibilidade.png', dpi=150)
plt.close()
```

## 7. Geometric Interpretation

```
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(111, projection='3d')
xx, yy = np.meshgrid(np.linspace(-2, 6, 20), np.linspace(-2, 6, 20))
z1 = (20 - 2*xx - 3*yy)
z2 = (22 - xx - 2*yy) / 3
z3 = (18 - 3*xx - yy) / 2
ax.plot_surface(xx, yy, z1, alpha=0.3)
ax.plot_surface(xx, yy, z2, alpha=0.3)
ax.plot_surface(xx, yy, z3, alpha=0.3)
ax.scatter(x_gauss[0], x_gauss[1], x_gauss[2], color='red', s=50)
plt.savefig('assets/figures/intersecao_planos.png', dpi=150)
plt.close()
```
