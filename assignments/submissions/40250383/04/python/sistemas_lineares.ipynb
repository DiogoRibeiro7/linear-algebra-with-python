```python
import numpy as np

A = np.array([[2,1],
              [1,3]])

b = np.array([8,13])

x = np.linalg.solve(A, b)

print("Solução:")
print(x)
```

```python
A = np.array([[2,1,8],
              [1,3,13]], dtype=float)

linhas = len(A)

for i in range(linhas):

    A[i] = A[i] / A[i,i]

    for j in range(linhas):
        if i != j:
            A[j] = A[j] - A[j,i] * A[i]

print("Matriz reduzida:")
print(A)
```

```python
A1 = np.array([[1,1,2],
               [2,2,5]], dtype=float)

A2 = np.array([[1,1,2],
               [2,2,4]], dtype=float)

print("Sistema impossível:")
print(A1)

print("\nSistema com infinitas soluções:")
print(A2)
```

```python
A = np.array([[2,1],
              [1,3]])

b = np.array([8,13])

x = np.linalg.solve(A, b)

print("Solução:")
print(x)

print("\nVerificação:")
print(A @ x)
```
