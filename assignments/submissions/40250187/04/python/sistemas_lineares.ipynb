```python
import numpy as np

print("=" * 50)
print("T4 — Propriedades da Transposta")
print("=" * 50)

# Matrizes
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [6, 5, 4],
    [3, 2, 1]
])

C = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

# -----------------------------------
# (Aᵀ)ᵀ = A
# -----------------------------------

prop1 = np.allclose((A.T).T, A)

print("\n1) (Aᵀ)ᵀ = A")
print(prop1)

# -----------------------------------
# (A + B)ᵀ = Aᵀ + Bᵀ
# -----------------------------------

lado_esquerdo_2 = (A + B).T
lado_direito_2 = A.T + B.T

prop2 = np.allclose(
    lado_esquerdo_2,
    lado_direito_2
)

print("\n2) (A + B)ᵀ = Aᵀ + Bᵀ")
print(prop2)

# -----------------------------------
# (AC)ᵀ = CᵀAᵀ
# -----------------------------------

lado_esquerdo_3 = (A @ C).T
lado_direito_3 = C.T @ A.T

prop3 = np.allclose(
    lado_esquerdo_3,
    lado_direito_3
)

print("\n3) (AC)ᵀ = CᵀAᵀ")
print(prop3)

# Mostrar resultados
print("\n(AC)ᵀ:")
print(lado_esquerdo_3)

print("\nCᵀAᵀ:")
print(lado_direito_3)
```
