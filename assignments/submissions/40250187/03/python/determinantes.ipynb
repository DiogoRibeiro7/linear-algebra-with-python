```python
import numpy as np

print("=" * 50)
print("T3 — Não comutatividade e comutador")
print("=" * 50)

# Matrizes 2x2
P = np.array([
    [1, 2],
    [0, 1]
])

Q = np.array([
    [2, 0],
    [1, 3]
])

# Produtos
PQ = P @ Q
QP = Q @ P

# Comutador
comutador = PQ - QP

# Matriz identidade
I = np.eye(2)

print("Matriz P:")
print(P)

print("\nMatriz Q:")
print(Q)

print("\nPQ:")
print(PQ)

print("\nQP:")
print(QP)

print("\nComutador [P,Q] = PQ - QP:")
print(comutador)

# Verificação
print("\nPQ é igual a QP?")
print(np.allclose(PQ, QP))

# Identidade
print("\nP @ I:")
print(P @ I)

print("\nI @ P:")
print(I @ P)

print("\nA identidade preserva P?")
print(
    np.allclose(P @ I, P)
    and np.allclose(I @ P, P)
)
```
