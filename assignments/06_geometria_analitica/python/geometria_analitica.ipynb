# Assignment 6 — Analytic Geometry: Lines and Planes
This notebook explores lines and planes with 2D and 3D visualizations.

## 1. Lines in R²

```
import numpy as np
import matplotlib.pyplot as plt
from linalg_utils.geometry import posicao_relativa_retas_2d

pairs = [
    ('Intersecting', np.array([0, 0]), np.array([1, 1]), np.array([0, 1]), np.array([1, -1])),
    ('Parallel', np.array([0, 0]), np.array([1, 0]), np.array([0, 1]), np.array([1, 0])),
    ('Coincident', np.array([0, 0]), np.array([1, 1]), np.array([1, 1]), np.array([2, 2])),
]

fig, axes = plt.subplots(1, 3, figsize=(12, 3))
t = np.linspace(-2, 2, 200)
for ax, (label, p1, d1, p2, d2) in zip(axes, pairs):
    x1 = p1[0] + t * d1[0]
    y1 = p1[1] + t * d1[1]
    x2 = p2[0] + t * d2[0]
    y2 = p2[1] + t * d2[1]
    ax.plot(x1, y1, label='Line 1')
    ax.plot(x2, y2, label='Line 2')
    ax.set_title(posicao_relativa_retas_2d(p1, d1, p2, d2))
    ax.legend()
plt.tight_layout()
plt.savefig('assets/figures/retas_2d.png', dpi=150)
plt.close()
```

## 2. Lines in R³

```
from linalg_utils.geometry import posicao_relativa_retas_3d
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

pairs = [
    ('Intersecting', np.array([0, 0, 0]), np.array([1, 1, 1]), np.array([1, 0, 0]), np.array([-1, 1, 0])),
    ('Parallel', np.array([0, 0, 0]), np.array([1, 0, 0]), np.array([0, 1, 0]), np.array([1, 0, 0])),
    ('Coincident', np.array([0, 0, 0]), np.array([1, 2, 3]), np.array([1, 2, 3]), np.array([2, 4, 6])),
    ('Skew', np.array([0, 0, 0]), np.array([1, 0, 0]), np.array([0, 1, 1]), np.array([0, 1, 0])),
]

fig = plt.figure(figsize=(8, 8))
for i, (label, p1, d1, p2, d2) in enumerate(pairs, start=1):
    ax = fig.add_subplot(2, 2, i, projection='3d')
    t = np.linspace(-2, 2, 200)
    r1 = p1.reshape(3, 1) + d1.reshape(3, 1) * t
    r2 = p2.reshape(3, 1) + d2.reshape(3, 1) * t
    ax.plot(r1[0], r1[1], r1[2], label='Line 1')
    ax.plot(r2[0], r2[1], r2[2], label='Line 2')
    ax.set_title(posicao_relativa_retas_3d(p1, d1, p2, d2))
    ax.legend()
plt.tight_layout()
plt.savefig('assets/figures/retas_3d.png', dpi=150)
plt.close()
```

## 3. Planes — Representations

```
from linalg_utils.geometry import plano_cartesiano_para_parametrico

planes = [(2, 3, -1, 6), (1, -2, 2, 4)]
for a, b, c, d in planes:
    p, v1, v2 = plano_cartesiano_para_parametrico(a, b, c, d)
    print(f'Plane: {a}x + {b}y + {c}z = {d}')
    print('Parametric form: P + s*v1 + t*v2')
    print('P =', p, 'v1 =', v1, 'v2 =', v2)
    print('Normal:', np.array([a, b, c]))
    print('-' * 40)
```

## 4. Relative Position of Planes

```
from linalg_utils.geometry import posicao_relativa_planos

pairs = [
    ('Secant', (1, 1, 1, 1), (1, -1, 1, 0)),
    ('Parallel', (1, 1, 1, 1), (1, 1, 1, 3)),
    ('Coincident', (1, 1, 1, 1), (2, 2, 2, 2)),
]

for label, p1, p2 in pairs:
    n1 = np.array(p1[:3]); d1 = p1[3]
    n2 = np.array(p2[:3]); d2 = p2[3]
    print(label, '->', posicao_relativa_planos(n1, d1, n2, d2))
```

## 5. Line-Plane Intersection

```
from linalg_utils.geometry import intersecao_reta_plano

plane_normal = np.array([2, 3, 1])
plane_d = 6

lines = [
    ('Intersecting', np.array([1, 1, 1]), np.array([1, 0, 0])),
    ('Parallel', np.array([0, 0, 0]), np.array([3, -2, -6])),
    ('Contained', np.array([3, 0, 0]), np.array([-3, 2, 0])),
]

for label, p, d in lines:
    inter = intersecao_reta_plano(p, d, plane_normal, plane_d)
    print(label, '->', inter)
```

## 6. Parallelism and Perpendicularity

```
d1 = np.array([1, 0, 0])
d2 = np.array([0, 1, 0])
n1 = np.array([0, 0, 1])
n2 = np.array([1, 0, 0])

print('d1 · d2 =', np.dot(d1, d2), '(perpendicular lines)')
print('d1 · n1 =', np.dot(d1, n1), '(line parallel to plane with normal n1)')
print('n1 · n2 =', np.dot(n1, n2), '(perpendicular planes if 0)')
```

## 7. Complete Visual Summary

```
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(111, projection='3d')

xx, yy = np.meshgrid(np.linspace(-2, 2, 20), np.linspace(-2, 2, 20))
z = (6 - 2*xx - 3*yy)
ax.plot_surface(xx, yy, z, alpha=0.4)

t = np.linspace(-2, 2, 100)
line1 = np.array([1, 1, 1])[:, None] + np.array([1, 0, 0])[:, None] * t
line2 = np.array([0, 0, 2])[:, None] + np.array([1, 0, 0])[:, None] * t
ax.plot(line1[0], line1[1], line1[2], label='Intersecting line')
ax.plot(line2[0], line2[1], line2[2], label='Parallel line')

ax.quiver(0, 0, 0, 2, 3, 1, length=1, color='red', label='Normal')
ax.scatter([1], [1], [1], color='black')
ax.legend()
plt.savefig('assets/figures/geometria_resumo.png', dpi=150)
plt.close()
```
