## Task 1 — simple solve
Solve `A x = b` for a well-conditioned `A` and print the solution.


```python
import numpy as np
A = np.array([[3.0, 1.0], [1.0, 2.0]])
b = np.array([9.0, 8.0])
# TODO: solve and print
x = np.linalg.solve(A, b)
print('Solution:', x)
```

## Task 2 — residual
Compute the residual `Ax - b` and describe its norm.


```python
residual = A @ x - b
# TODO: report norm
print('Residual norm:', np.linalg.norm(residual))
```

Use `np.linalg.cond` to inspect the matrix's condition number and note whether it's well-conditioned.


```python
# TODO: compute condition number
print('Condition number:', np.linalg.cond(A))
```

## Task 4 — perturbed RHS
Perturb `b` slightly and solve again. Compare the change in solution to the perturbation size.


```python
delta = 1e-3
b_perturbed = b + np.array([delta, -delta])
# TODO: solve and compare
x_pert = np.linalg.solve(A, b_perturbed)
print('Perturbed change:', np.linalg.norm(x_pert - x))
```
