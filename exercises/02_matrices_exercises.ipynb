## Task 1 — build and apply a transform
Create a matrix that rotates by 30° and scales the x-axis by 1.2. Apply it to `[1, 0.5]` and print both vectors.


```python
import numpy as np
from linalg_with_python import geometry2d

# TODO: define your transformation matrix
transform = np.eye(2)
v = np.array([1.0, 0.5])
mapped = transform @ v
print('Original:', v)
print('Mapped:', mapped)
```

## Task 2 — basis vectors
Use `LinearMap2D` to map `e1` and `e2`. Explain whether the transformed basis is still orthogonal.


```python
lm = geometry2d.LinearMap2D(transform)
basis = np.eye(2)
# TODO: map each basis vector and inspect dot products
mapped_basis = geometry2d.apply_linear_map(transform, basis)
print('DOT product between mapped e1 and e2:', np.dot(mapped_basis[0], mapped_basis[1]))
```

## Task 3 — unit square boundary
Generate `unit_square(120)` and plot both the original and transformed polygons.


```python
square = geometry2d.unit_square(120)
# TODO: apply `geometry2d.apply_linear_map` and visualize the polygons
square_mapped = geometry2d.apply_linear_map(transform, square)
```

## Task 4 — unit circle
Map the `unit_circle()` through your matrix and report the max distance between transformed points.


```python
circle = geometry2d.unit_circle()
circle_mapped = geometry2d.apply_linear_map(transform, circle)
# TODO: compute how far transformed points stray from the origin
distances = np.linalg.norm(circle_mapped, axis=1)
print('Max radius:', distances.max())
```

## Task 5 — composition
Compose your matrix with a shear (k=0.5) and invert the result. What happens to the unit square?


```python
shear = np.array([[1.0, 0.5], [0.0, 1.0]])
composed = shear @ transform
# TODO: invert `composed` and apply to the mapped square
inv_composed = np.linalg.inv(composed)
recovered = geometry2d.apply_linear_map(inv_composed, square_mapped)
print('Recovered first vertex:', recovered[0])
```

## Task 6 — eigen-direction
Guess a vector that might be stretched the most, apply the transform, and compare length ratios.


```python
candidate = np.array([1.0, 1.0])
mapped_candidate = geometry2d.apply_linear_map(transform, candidate[np.newaxis, :])[0]
# TODO: compute stretching factor
print('Stretch factor:', np.linalg.norm(mapped_candidate) / np.linalg.norm(candidate))
```
