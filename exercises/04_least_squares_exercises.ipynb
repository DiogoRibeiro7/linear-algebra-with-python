# Least squares exercises
Apply both normal equations and QR/Gram-Schmidt solvers, compare residuals, and plot the fits.


## Task 1 — design matrix
Create a Vandermonde-style design matrix for data `x = [0, 1, 2, 3]` and `y = [1, 3, 7, 13]`.


```python
import numpy as np
from linalg_with_python import least_squares

x = np.array([0.0, 1.0, 2.0, 3.0])
y = np.array([1.0, 3.0, 7.0, 13.0])
# TODO: build matrix [x, ones(x)]
A = np.vstack([x, np.ones_like(x)]).T
```

## Task 2 — normal equations
Use `(A.T @ A) @ x = A.T @ y` to fit and note residual norm.


```python
normal_x = np.linalg.inv(A.T @ A) @ A.T @ y
# TODO: compute residual
residual_normal = np.linalg.norm(A @ normal_x - y)
print('Normal residual:', residual_normal)
```

## Task 3 — QR solver
Use `least_squares.least_squares_qr` and compare the `residual_norm` to the normal equation.


```python
qr_result = least_squares.least_squares_qr(A, y)
# TODO: report result
print('QR residual:', qr_result.residual_norm)
```

## Task 4 — compare coefficients
Print out the fitted coefficients from both methods and look for subtle differences.


```python
print('Normal coeffs:', normal_x)
print('QR coeffs:', qr_result.x)
```

## Task 5 — plot data + fit
Use matplotlib to plot the points plus the QR fit line.


```python
import matplotlib.pyplot as plt
plt.plot(x, y, 'o', label='data')
plt.plot(x, A @ qr_result.x, '-', label='QR fit')
# TODO: add labels/title
plt.title('Least squares fit')
plt.legend()
```

## Task 6 — projection residual
Compute `projected = A @ qr_result.x` and inspect whether `y - projected` is orthogonal to `A`'s columns.


```python
proj = A @ qr_result.x
res = y - proj
# TODO: check orthogonality with columns
print('Dot with first col:', np.dot(res, A[:, 0]))
```

## Task 7 — relative error
Use `least_squares.RelativeError` (or compute manually) to compare normal vs QR results.


```python
relative_error = np.linalg.norm(normal_x - qr_result.x) / np.linalg.norm(qr_result.x)
# TODO: interpret this number
print('Relative diff:', relative_error)
```

## Task 8 — reflection
What happens if you reflect the data across the vertical axis before fitting? Does QR still win?


```python
A_reflect = A.copy()
A_reflect[:, 0] *= -1
qr_reflect = least_squares.least_squares_qr(A_reflect, y)
# TODO: compare residuals to previous ones
print('Reflected residual:', qr_reflect.residual_norm)
```

## Check your work

- The projection residual should be orthogonal to each column of `A`.
- Relative differences near machine tolerance indicate consistent results.
- Reflection should only affect the sign/drop of coefficients, not the residual magnitude.
